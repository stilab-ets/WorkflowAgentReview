```python
import pandas as pd
commit_details = pd.read_parquet('../datasets/aidev/pr_commit_details.parquet')
fetched_commits =  pd.read_csv('../datasets/processed/fetched_eligible_pop_commits.csv')
aidev_pr =  pd.read_parquet('../datasets/aidev/pull_request.parquet')

```

```python
fetched_commits = fetched_commits.sort_values('commit.author.date')
fetched_commits
```

# Extract workflow files

```python
workflow_commit_files = (
    commit_details[
        commit_details["filename"].str.startswith(".github/workflows/", na=False)
        &
        commit_details["filename"].str.endswith((".yaml", ".yml"), na=False)
    ]

)
print(f"Number of commit details that touch workflow files: {len(workflow_commit_files.groupby(['pr_id', 'sha']).size())}")
workflow_commit_files

```

# Extract workflow commits

```python
workflow_commits = workflow_commit_files[['pr_id','sha','filename','status','changes','patch']].merge(
    fetched_commits,
    on=["pr_id", "sha"],
    how="inner"
)

print(len(workflow_commits))
workflow_commits
```

# Extract workflow First commit

```python
workflow_commits = workflow_commits.sort_values('commit.author.date')
first_workflow_commit = workflow_commits.loc[
    workflow_commits.groupby(["pr_id","pr_url"])["commit.author.date"].idxmin(),
    
]

# exclude any commit has empty code 
first_workflow_commit= first_workflow_commit[~first_workflow_commit['patch'].isna()]
# first_workflow_commit[first_workflow_commit['status']=='removed']
first_workflow_commit
```

# Extract PRs with workflow changes in the first commit (eligible)

```python
eligible_prs = aidev_pr.merge(first_workflow_commit[['pr_url','sha']],left_on='html_url',right_on='pr_url')
eligible_prs = eligible_prs[list(aidev_pr.columns)+['sha']]
eligible_prs = eligible_prs.rename(columns={"sha":"first_commit_sha"})
eligible_prs = eligible_prs[eligible_prs['state']=='closed']
eligible_prs= eligible_prs.rename(columns={'id':'pr_id'})
eligible_prs.to_csv('../datasets/processed/ai_eligible_prs.csv', index=False)
eligible_prs 
```

```python
eligible_prs['repo_url'].nunique()
```

# Eligible AI commits

```python
eligible_prs['pr_url'] = eligible_prs['html_url']
eligible_ai_commits = eligible_prs[['pr_url','pr_id']].merge(fetched_commits,on=['pr_url','pr_id'])
eligible_ai_commits.to_csv('../datasets/processed/ai_eligible_fetched_commits.csv', index=False)
eligible_ai_commits
```

# Extract ai eligible commit details

```python
eligible_commit_details = eligible_ai_commits[['pr_id','sha']].merge(commit_details,on=['pr_id','sha'])
eligible_commit_details.to_csv('../datasets/processed/ai_eligible_commit_details.csv', index=False)
eligible_commit_details
```

# Eligible reviews

```python
# extract eligible reviews 
ai_reviews = pd.read_parquet('../datasets/aidev/pr_reviews.parquet')
eligible_ai_reviews = eligible_prs[['pr_id']].merge(ai_reviews,on='pr_id')
eligible_ai_reviews.to_csv('../datasets/processed/ai_eligible_reviews.csv', index=False)
eligible_ai_reviews
```

# Eligible review comments

```python
review_comments = pd.read_parquet('../datasets/aidev/pr_review_comments_v2.parquet')
eligible_review_comments = eligible_ai_reviews[['id']].rename(columns={'id': 'pull_request_review_id'}).merge(review_comments,on='pull_request_review_id')
eligible_review_comments.to_csv('../datasets/processed/ai_eligible_review_comments.csv', index=False)
eligible_review_comments
```

# Eligible ai comments

```python
ai_comments = pd.read_parquet('../datasets/aidev/pr_comments.parquet')
eligible_ai_comments = eligible_prs[['pr_id']].merge(ai_comments,on='pr_id')
eligible_ai_comments.to_csv('../datasets/processed/ai_eligible_comments.csv', index=False)
eligible_ai_comments
```

# Eligible builds

## Get raw data

```python
builds_raw = pd.read_json('../datasets/raw/eligible_commits_check_runs.jsonl', lines=True)
ai_build_raw = builds_raw[builds_raw['pr_type']=='ai']
ai_build_raw
```

## Filter builds , eligible commits only

### Check eligible commits in fetched build commits

```python
#Filter build to include build of eligible pr commit's
print(f"eligible commits {len(eligible_ai_commits)}")
print(f"fetched build commits {len(ai_build_raw)}")
print(f"fetched eligible builds commits {len(ai_build_raw[ai_build_raw['commit'].isin(eligible_ai_commits['sha'])])}")
```

### Filter build commits by eligible commits

```python
eligible_builds_raw = ai_build_raw[ai_build_raw['commit'].isin(eligible_ai_commits['sha'])]
print(f"eligible builds raw {len(eligible_builds_raw)}")
eligible_builds_raw
```

## Extract Build data

```python
eligible_builds_raw_normalized = pd.json_normalize(eligible_builds_raw['check_runs'])
eligible_builds_raw_normalized = pd.concat([eligible_builds_raw.drop(columns=['check_runs']), eligible_builds_raw_normalized], axis=1)

eligible_builds_raw_exploded = eligible_builds_raw_normalized.explode('check_runs')
eligible_builds_raw_exploded['check_runs'] = eligible_builds_raw_exploded['check_runs'].apply(lambda x: x if isinstance(x, dict) else {})
eligible_builds_normalized = pd.json_normalize(eligible_builds_raw_exploded['check_runs'])
eligible_builds = pd.concat([eligible_builds_raw_exploded.drop(columns=['check_runs']), eligible_builds_normalized], axis=1)
eligible_builds
```
